```python
class Nodo:
    def __init__(self,t):
        self.hijos = list()
        self.llaves = list()
        self.hoja=1
        self.n=0
        for k in range(2*t):
            self.llaves.append([None])
        for k in range(2*t+1):
            self.hijos.append([None])

class ArbolB:
    def __init__(self,gradoMinimo):
        self.t= gradoMinimo
        self.raiz = None
    
    def bTreeCreate(self):
        if(self.raiz==None):
            self.raiz=Nodo(self.t)
        return self.raiz

    def bTreeSplitShild(self,x,i):
        z=Nodo(self.t)
        y = x.hijos[i]
        z.hoja=y.hoja
        z.n=self.t-1 
        
        for j in range(1,self.t):
            z.llaves[j]=y.llaves[j+self.t]
            y.llaves[j+self.t]=None
                
        if y.hoja==0:
            for j in range(1,self.t+1):
                z.hijos[j]=y.hijos[j+self.t]
                y.hijos[j+self.t]=None
        y.n=self.t-1
        for j in range(x.n+1,i,-1):
            x.hijos[j+1]=x.hijos[j]
            
        x.hijos[i+1]=z
        
        for j in range (x.n,i-1,-1):
            x.llaves[j+1]=x.llaves[j]
        x.llaves[i]=y.llaves[self.t]
        y.llaves[self.t]=None
        x.n=x.n+1
        
        
    def bTreeInsertNonFull(self,x,k):
        i=x.n
        if x.hoja == 1:
            while ( i >= 1) and (k < x.llaves[i]):
                x.llaves[i+1]= x.llaves[i]
                i=i-1
            x.llaves[i+1]=k
            x.n=x.n+1
            
             #escribir a disco
        else:
            #No es hoja
            while (i >= 1) and (k < x.llaves[i]):
                i=i-1
            i=i+1
                #leer disco
            if x.hijos[i].n == 2*self.t-1:
                self.bTreeSplitShild(x,i)
                if k > x.llaves[i]:
                    i=i+1
            self.bTreeInsertNonFull(x.hijos[i],k)

    def bTreeInsert(self,nodo, k):
        r=self.raiz
         #nodo lleno
        if r.n == 2*self.t-1:
            s=Nodo(self.t)
            self.raiz=s
            s.hoja=0
            s.n=0
            s.hijos[1]=r
            self.bTreeSplitShild(s,1)
            self.bTreeInsertNonFull(s,k)
           
        else:
            self.bTreeInsertNonFull(r,k)

    def imprimeNodo(self,nodo):
        for i in range(1,2+self.t,1):
            a=nodo.llaves[i]
            if(a != [None] and a != None):
                print(chr(nodo.llaves[i]))
    def bTreeSearch(self,nodo,llave):
        i=1
        while(i<=nodo.n and llave > nodo.llaves[i]):
            i+=1
        if i<=nodo.n and llave == nodo.llaves[i]:
            print("Si fue encontrado:",chr(nodo.llaves[i]))
            return nodo,i
        else:
            if nodo.hoja==1:
                return None,None
            else:
                return self.bTreeSearch(nodo.hijos[i],llave)

def controladora():
    BT = ArbolB(2) 
    ra=BT.bTreeCreate()
    print("Inserta B, T y H")
    BT.bTreeInsert(ra, ord("B"))
    BT.bTreeInsert(ra, ord("T"))
    BT.bTreeInsert(ra, ord("H"))

    print("Raiz")
    BT.imprimeNodo(ra)

    print("Insertar M, O y C")
    BT.bTreeInsert(ra, ord("M"))
    BT.bTreeInsert(ra, ord("O"))
    BT.bTreeInsert(ra, ord("C"))


    print("Impresion hasta insertar M, O y C")
    print("Raiz")
    BT.imprimeNodo(BT.raiz)
    print("Hijo 1")
    BT.imprimeNodo(BT.raiz.hijos[1])
    print("Hijo 2")
    BT.imprimeNodo(BT.raiz.hijos[2])


    print("Insertar Z, G y L")
    BT.bTreeInsert(BT.raiz, ord("Z"))
    BT.bTreeInsert(BT.raiz, ord("G"))
    BT.bTreeInsert(BT.raiz, ord("L"))


    print("Impresion hasta insertar Z, G y L")
    print("Raiz")
    BT.imprimeNodo(BT.raiz)
    print("Hijo 1")
    BT.imprimeNodo(BT.raiz.hijos[1])
    print("Hijo 2")
    BT.imprimeNodo(BT.raiz.hijos[2])
    print("Hijo 3")
    BT.imprimeNodo(BT.raiz.hijos[3])

    print("Insertar E")
    BT.bTreeInsert(BT.raiz, ord("E"))

    print("Impresion hasta insertar E")
    print("Raiz")
    BT.imprimeNodo(BT.raiz)
    print("Hijo 1")
    BT.imprimeNodo(BT.raiz.hijos[1])
    print("Hijo 2")
    BT.imprimeNodo(BT.raiz.hijos[2])
    print("Hijo 3")
    BT.imprimeNodo(BT.raiz.hijos[3])
    print("Hijo 4")
    BT.imprimeNodo(BT.raiz.hijos[4])

    print("Insertar N")
    BT.bTreeInsert(BT.raiz, ord("N"))

    print("Impresion hasta insertar N")
    print("Raiz")
    BT.imprimeNodo(BT.raiz)
    print("Hijo 1")
    BT.imprimeNodo(BT.raiz.hijos[1])
    print("Hijo 2")
    BT.imprimeNodo(BT.raiz.hijos[2])

    print("Hijo 1 de hijo 1")
    BT.imprimeNodo(BT.raiz.hijos[1].hijos[1])
    print("Hijo 2 de hijo 1")
    BT.imprimeNodo(BT.raiz.hijos[1].hijos[2])
    print("Hijo 1 de hijo 2")
    BT.imprimeNodo(BT.raiz.hijos[2].hijos[1])
    print("Hijo 2 de hijo 2")
    BT.imprimeNodo(BT.raiz.hijos[2].hijos[2])

    nodo,i=BT.bTreeSearch(BT.raiz, ord("M"))
    print(str(nodo.llaves[i]), i)

    
    
controladora()

        
```
